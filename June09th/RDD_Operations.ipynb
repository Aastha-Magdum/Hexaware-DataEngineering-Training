```python
from pyspark.sql import SparkSession

spark=SparkSession.builder \
.appName("RDD") \
.getOrCreate()
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
print(rdd.collect())
```

```python
rdd.count()
```

```python
rdd.take(3)
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
res=rdd.map(
    lambda x:x*10
)

print(res.collect())
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
res=rdd.map(
    lambda x: x*x

)
print(res.collect())
```

```python
numbers=[1,2,3,4,5,6,7,8]
rdd=spark.sparkContext.parallelize(numbers)
res=rdd.filter(
    lambda x: x % 2==0

)
print(res.collect())
```

```python
numbers=[10,20,30,40,50,30,20]
rdd=spark.sparkContext.parallelize(numbers)
res=rdd.distinct()

print(res.collect())
```

```python
rdd1=spark.sparkContext.parallelize([10,20,30])
rdd2=spark.sparkContext.parallelize([40,50,60])
res=rdd1.union(rdd2)
print(res.collect())
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
res=rdd.sortBy(
    lambda x: x
)
print(res.collect())
```

```python
numbers=[10,20,30,40,50]
rdd=spark.sparkContext.parallelize(numbers)
res=rdd.reduce(
    lambda x,y:x-y
)
print(res)
```
