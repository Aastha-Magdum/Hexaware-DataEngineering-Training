```python
#91
doctors_df = spark.read.option("header", True).option("inferSchema", True).csv("doctors.csv")

visits_df = spark.read.option("header", True).option("inferSchema", True).csv("visits.csv")

hospitals_df = spark.read.option("multiline", True).json("hospital_config.json")
```

```python
#92
visits_clean_df = visits_df.fillna({"bill_amount": 0})
```

```python
#93
hospital_flat_df = hospitals_df.select(
    "hospital_id",
    "hospital_name",
    "city",
    col("contact.phone").alias("phone"),
    col("contact.email").alias("email"),
    "services"
)

hospital_flat_df.show(truncate=False)
```

```python
#94
doctor_visit_df = doctors_df.join(
    visits_clean_df,
    "doctor_id",
    "inner"
)

doctor_visit_df.show()
```

```python
#95
revenue_df = doctor_visit_df.withColumn(
    "revenue",
    col("bill_amount")
)

revenue_df.show()
```

```python
#96
doctor_rank_df = revenue_df.groupBy(
    "doctor_id",
    "doctor_name"
).agg(
    sum("revenue").alias("total_revenue")
)

window_spec = Window.orderBy(col("total_revenue").desc())

doctor_rank_df.withColumn(
    "rank",
    dense_rank().over(window_spec)
).show()
```

```python
#97
revenue_df.groupBy("specialization") \
    .agg(
        sum("revenue").alias("total_revenue"),
        avg("revenue").alias("avg_revenue")
    ) \
    .show()
```

```python
#98
revenue_df.write.mode("overwrite").parquet("silver_hospital_data")
```

```python
#99
gold_report_df = revenue_df.groupBy(
    "specialization"
).agg(
    sum("revenue").alias("total_revenue")
)

gold_report_df.show()
```

```python
#100
dashboard_df = revenue_df.groupBy(
    "city",
    "specialization"
).agg(
    count("visit_id").alias("total_visits"),
    sum("revenue").alias("total_revenue")
)

dashboard_df.show()
```
