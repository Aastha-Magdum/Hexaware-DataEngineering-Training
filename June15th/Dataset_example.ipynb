```python
from pyspark.sql import SparkSession

spark=SparkSession.builder \
.appName("Datasetex") \
.getOrCreate()
```

```python
data = [
    (101, "Rahul Sharma", "Hyderabad", "Electronics", 65000),
    (102, "Priya Reddy", "Bangalore", "Electronics", 25000),
    (103, "Amit Kumar", "Mumbai", "Furniture", 12000),
    (104, "Sneha Patel", "Chennai", "Fashion", 8000),
    (105, "Farhan Ali", "Delhi", "Electronics", 45000)
]

columns = [
    "order_id",
    "customer_name",
    "city",
    "category",
    "amount"
]

df = spark.createDataFrame(data, columns)

display(df)
```
