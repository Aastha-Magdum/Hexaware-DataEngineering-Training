```python
df=spark.read.csv(
    "flightBookings.csv",
    header=True,
    inferSchema=True
)
df.show()
```

```python
#1
res=df
res.show()
```

```python
#2
res=df.select("passenger_name","from_city","to_city")
res.show()
```

```python
#3
res=df.filter(df.status=="Confirmed")
res.show()
```

```python
#4
res=df.filter(df.status=="Cancelled")
res.show()
```

```python
#5
res = df.filter(df.status == "Pending")
res.show()
```

```python
#6
res=df.filter(df.ticket_price>20000)
res.show()
```

```python
#7
res=df.filter(df.to_city.isin("Dubai","Singapore","London"))
res.show()
```

```python
#8
print(df.count())
```

```python
#9
res=df.groupby("airline").count()
res.show()
```

```python
#10
res=df.groupby("status").count()
res.show()
```

```python
#11
from pyspark.sql.functions import sum
res=df.filter(df.status=="Confirmed")
res.select(sum("ticket_price")).show()
```

```python
#12
from pyspark.sql.functions import avg
res=df.groupby("airline").avg("ticket_price")
res.show()
```

```python
#13
from pyspark.sql.functions import max
res=df.select(max("ticket_price"))
res.show()
```

```python
#14
from pyspark.sql.functions import min

res=df.select(min("ticket_price"))
res.show()
```

```python
#15
res=df.orderBy(df.ticket_price.desc())
res.show()
```

```python
from pyspark.sql.functions import col
res=df.withColumn("tax",col("ticket_price")*0.05)
res.show()
```

```python
#17
from pyspark.sql.functions import col
res=df.withColumn("tax",col("ticket_price")*0.05)
res=res.withColumn("final_price",col("ticket_price")+ col("tax"))
res.show()
```

```python
from pyspark.sql.functions import when
res=df.withColumn(
    "price_category",
    when(df.ticket_price>=30000,"Premium")
    .when(df.ticket_price>=10000,"Standard")
    .otherwise("Budget")
)
res.show()
```

```python
from pyspark.sql.functions import when
res=df.withColumn(
    "price_category",
    when(df.ticket_price>=30000,"Premium")
    .when(df.ticket_price>=10000,"Standard")
    .otherwise("Budget")
)
res.groupBy("price_category").count().show()
```

```python
#20
res=df.filter(df.status=="Confirmed")
res.write.mode("overwrite").parquet("confirmed_flights.parquet")
```

```python
hotels_df = spark.read.option(
"multiline",
"true"
).json("hotels.json")
hotels_df.show(truncate=False)
hotels_df.printSchema()
```

```python
#1
hotels_df.show()
```

```python
#2
res = hotels_df.select("hotel_name", "city", "rating")
res.show()
```

```python
#3
res = hotels_df.filter(hotels_df.category == "Luxury")
res.show()
```

```python
#4
res=hotels_df.filter(hotels_df.rating>4.5)
res.show()
```

```python
#5
res = hotels_df.filter(hotels_df.rooms_available > 15)
res.show()
```

```python
#6
res = hotels_df.filter(hotels_df.price_per_night > 10000)
res.show()
```

```python
#7
res = hotels_df.filter(hotels_df.city.isin("Dubai", "London"))
res.show()
```

```python
#8
res = hotels_df.filter(hotels_df.contact.phone.isNull())
res.show()
```

```python
#9
res = hotels_df.filter(hotels_df.contact.email.isNull())
res.show()
```

```python
#10
res = hotels_df.select(
    "hotel_name",
    "contact.phone",
    "contact.email"
)
res.show()
```

```python
#11
from pyspark.sql.functions import array_contains

res = hotels_df.filter(
    array_contains("amenities", "wifi")
)
res.show()
```

```python
#12
from pyspark.sql.functions import array_contains

res = hotels_df.filter(
    array_contains("amenities", "spa")
)
res.show()
```

```python
#13
res = hotels_df.groupBy("city").count()
res.show()
```

```python
#14
res = hotels_df.groupBy("category").count()
res.show()
```

```python
#15
from pyspark.sql.functions import avg

res = hotels_df.groupBy("category").avg("rating")
res.show()
```

```python
#16
from pyspark.sql.functions import avg

res = hotels_df.groupBy("city").avg("price_per_night")
res.show()
```

```python
#17
from pyspark.sql.functions import max

res = hotels_df.select(max("price_per_night"))
res.show()
```

```python
#18
from pyspark.sql.functions import col

res = hotels_df.withColumn(
    "total_potential_revenue",
    col("rooms_available") * col("price_per_night")
)

res.show()
```

```python
#19
res = hotels_df.orderBy(hotels_df.rating.desc())
res.show()
```

```python
#20
res = hotels_df.select(
    "hotel_id",
    "hotel_name",
    "city",
    "category",
    "rating",
    "rooms_available",
    "price_per_night",
    "contact.phone",
    "contact.email"
)
res.write.mode("overwrite").parquet(
    "hotels_flattened.parquet"
)
```

```python
df=spark.read.csv(
    "Bookings.csv",
    header=True,
    inferSchema=True
)
df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import col

df.filter(col("city").isNull()).show()

df.filter(col("provider").isNull()).show()
```

```python
drop_any_null_df=df.na.drop()
drop_any_null_df.show()
```

```python
clean_df=df.na.drop(
    subset=[
        "customer_name",
        "service_type",
        "booking_amount"
    ]
)
clean_df.show()
```

```python
filled_df=df.na.fill(
  {
      "city":"Unknown"
  }
)
filled_df.show()
```

```python
filled_df=df.na.fill({
    "city":"Unknown",
    "provider":"Not Available",
    "booking_status":"Unknown",
    "payment_mode":"Not Available",
    "booking_amount":0
})
filled_df.show()
```
